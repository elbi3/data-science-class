## Part 2: Explore the Data

```python
# methods
# pick up at index for data cleaning to skip word-labels
def spanOfYearsRepresented(df, index=1):
    array = df.columns.values.to_numpy()
    for i, item in enumerate(array):
        if i < index:
            continue
        array[i] = item[0:4]
        # print(array[i])
    start = int(array[index])
    end = int(array[-2])
    print(end - start, "consecutive years,")
    print(array[index], " through ", array[-2])
```

## Part 4: Data Quality Assessment

Was not able to find missing values or duplicate rows. There were values of "0" however, which while being the correct data type indicate data absence. (No exports in some cases.) This could indicate exports were simply not recorded these years. If exports were not recorded for a significant number of years but exports did in fact happen, it may be wise to do metrics excluding these countries so that their impact on coffee production is not minimized.

Coffee production data has another idiosynchracy: instead of `int64` datatype, `float64` is used. It's unlikely that this adds anything to the dataset since digits right of the decimal are 0 (doublecheck this is 100% accurate across the dataset before taking action).

The production dataset had an odd indexing habit of using "year/year" instead of a single "year", as was done with imports and exports. Instead of "1990", production seemed to be calculated mid-year through next mid-year, "1990/91". This required a cleaning function to synchronize labeling.
