## Part 1: Quantifying Missing Data

```python
netflix_titles.isnull().sum()
```

```python
netflix_titles.isnull().mean() * 100
directorp = netflix_titles["director"].isnull().mean() * 100 # could also do (df.isnull().sum()/len(df)) * 100
castp = netflix_titles["cast"].isnull().mean() * 100
countryp = netflix_titles["country"].isnull().mean() * 100

print("| missing director: ", directorp, "| missing cast: ", castp, "| missing country: ", countryp, "|")
```

### Impact Scaling

While most columns have relatively few missing values, nearly 30% of the fields for "director" are missing, and almost 10% of the cast and country categories are missing. This dataset only provides 90% of cast and country data for movies, and a mere 70% of director information for the movie list. For a dataset with 8806 rows, 4 missing "ratings" and 3 missing "duration" is negligible.

## Part 2: Cleaning Strategies (Dropping vs. Imputing)

```python
# imputation
netflix_titles[["director"]] = netflix_titles[["director"]].fillna("Unknown")
netflix_titles["director"].isna().sum()
netflix_titles.isnull().sum()
```

```python
# row dropping
netflix_titles2.dropna(subset=["cast", "director"])
netflix_titles2.info
netflix_titles2.isnull().sum()
```

```python
netflix_titles2.shape
```

For the row-dropping strategy, our final shape was (5700, 12), down from (8807, 12). This is a los of 3107 rows, more than a third of the entire data set. There are many scenarios where this great a loss could ruin analysis: for instance, columns where we wanted to know metrics like "how many" or "to what extent" of movies over specific periods of time have certain characteristics. We would only be able to calculate data on 2/3 of the category.
