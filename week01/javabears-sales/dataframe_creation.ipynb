## 1. Construct the DataFrame

```python
# construct a DataFrame for coffee_sales
coffee_sales = {
    "Americano":[5628,6239],
    "Latte":[12426,25384],
    "Espresso":[852,756], 
    "Cappuccino":[8522,7561]}

df = pd.DataFrame(
    coffee_sales,
    index=["2021 Sales", "2022 Sales"])

df.index.name = "Year (Index)"

print(df)
```

## 2. Data Manipulation

```python
# (here I have calculated totals of individual columns)
americano = df["Americano"].sum()
latte = df["Latte"].sum()
espresso = df["Espresso"].sum()
cappuccino = df["Cappuccino"].sum()
print("total sales of Americano product: ", americano)
print("total sales of Latte product: ", latte)
print("total sales of Espresso product: ", espresso)
print("total sales of Cappuccino product: ", cappuccino)
```

### Calculate Total Product Volume

```python
total = americano + latte + espresso + cappuccino
print(total)
```

```python
# alternative way to calculate sum of entire df (requires not having a "Total" column yet, or count is doubled)
df.sum(numeric_only=True)
```

```python
# sum columns in a new row "Total" using .loc[] and referencing the specific columns in a list
df.loc["Total"] = df[["Americano", "Latte", "Espresso", "Cappuccino"]].sum()
print(df)
```

### Calculate Year-over-Year (YoY) Growth

```python
yoySalesAmericano = ((df["Americano"]["2022 Sales"] - df["Americano"]["2021 Sales"]) / df["Americano"]["2021 Sales"]) * 100
print(yoySalesAmericano)
```

```python
yoySalesLatte = ((df["Latte"]["2022 Sales"] - df["Latte"]["2021 Sales"]) / df["Latte"]["2021 Sales"]) * 100
print(yoySalesLatte)
```

```python
yoySalesEspresso = ((df["Espresso"]["2022 Sales"] - df["Espresso"]["2021 Sales"]) / df["Espresso"]["2021 Sales"]) * 100
print(yoySalesEspresso)
```

```python
yoySalesCappuccino = ((df["Cappuccino"]["2022 Sales"] - df["Cappuccino"]["2021 Sales"]) / df["Cappuccino"]["2021 Sales"]) * 100
print(yoySalesCappuccino)
```

```python
summary = pd.DataFrame(
    [yoySalesAmericano, yoySalesCappuccino, yoySalesEspresso, yoySalesLatte], 
    index=["yoySalesAmericano", "yoySalesCappuccino", "yoySalesEspresso", "yoySalesLatte"]).sort_values(by=0, ascending=False)

print(summary)
```

## 3. Business Insights & Strategy

### best-selling and least-selling beverage for both 2021 and 2022

```python
# 2021 data
x1 = df.loc["2021 Sales"]
max2021 = x1.max()
min2021 = x1.min()
index2021 = x1.idxmax()
indexMin2021 = x1.idxmin()

print("Best in 2021: ", index2021, max2021)
print("Worst in 2021: ", indexMin2021, min2021)
```

```python
# 2022 data

x2 = df.loc["2022 Sales"]
max2022 = x2.max()
min2022 = x2.min()
index2022 = x2.idxmax()
indexMin2022 = x2.idxmin()

print("Best in 2022: ", index2022, max2022)
print("Worst in 2022: ", indexMin2022, min2022)
```

### strategic analysis for JavaBears' leadership

For the period 2021-2022, our best selling item has been, consistently, the latte. Leadership might consider marketing creative variations of our best seller to capitalize on customer interest lattes. Drink "alternatives" such as "pumpkin spice" latte or "chai" latte require very little additional labor as well as a few low-cost ingredients. 

Our low-performer is espresso. While we have managed to sell over 25k units of our latte option, the espresso option has not broken the 1k mark. This seems to suggest a lack of customer interest, and it may be worth dropping the item from our "official" menu, altogether. Since the espresso only requires ingredients used in our other products, from a cost perspective, there is no need to effectively remove it from our products. Perhaps the marketing team can come up with a "secret menu" to make our espresso drink more appealing, and to explain the drink's absence. This level of secrecy may appeal to some customers while removing our need to do marketing to prop up the lackluster item.
