### 1. Standardizing country name strings so they match across all three data frames perfectly before merging.

```python
#check uniqueness of list
export_list = coffee_exp.iloc[:,0].is_unique
import_list = coffee_imp.iloc[:,0].is_unique
prod_list = coffee_prod.iloc[:,0].is_unique
print("EXPORTS: ", export_list, "  IMPORTS: ", import_list, "  PRODUCTION: ", prod_list)
```

```python
coffee_exp = coffee_exp.set_index("Country") #don't forget to assign the result back, or the new dataframe is lost
coffee_imp = coffee_imp.set_index("Country")
coffee_prod = coffee_prod.set_index("Country") 
coffee_exp
```

```python
# write an index cleaning function
def cleanUpCountryLabels(indx):
    newindx = indx.astype(str).str.strip().str.upper() #check for non-string types, stripe white space, change all to upper-case 
    return newindx

coffee_exp.index = cleanUpCountryLabels(coffee_exp.index)
coffee_imp.index = cleanUpCountryLabels(coffee_imp.index)
coffee_prod.index = cleanUpCountryLabels(coffee_prod.index)
```

```python
coffee_exp.index
coffee_imp.index
coffee_prod.index
```

### 2. Standardizing column names to int for range and calculations

```python
#checks
coffee_exp.columns.is_unique
coffee_exp.columns.dtype
coffee_exp.columns
#cleaning
coffee_exp.columns = coffee_exp.columns.str.strip()
years = pd.to_numeric(coffee_exp.columns[:-1])
coffee_exp.columns = list(years) + [coffee_exp.columns[-1]] #make years a list so that it joins the columns list
coffee_exp.columns
```

```python
#checks
coffee_imp.columns.is_unique
coffee_imp.columns.dtype
coffee_imp.columns
#cleaning
coffee_imp.columns = coffee_imp.columns.str.strip()
years = pd.to_numeric(coffee_imp.columns[:-1])
coffee_imp.columns = list(years) + [coffee_imp.columns[-1]]
coffee_imp.columns
```

```python
#checks
coffee_prod.columns.is_unique
coffee_prod.columns.inferred_type
coffee_prod.columns
#cleaning
coffee_prod.columns = coffee_prod.columns.str.strip() #only select year columns for range utility
years = pd.to_numeric(coffee_prod.columns[1:-1].str[:4]) #use .str to remove mismatched string endings: '1990/91' -> 1990
coffee_prod.columns = [coffee_prod.columns[0]] + list(years) + [coffee_prod.columns[-1]]
coffee_prod.columns
```
