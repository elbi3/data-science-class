```python
# write an index cleaning function
def cleanUpCountryLabels(indx):
    newindx = indx.astype(str).str.strip().str.upper() #check for non-string types, stripe white space, change all to upper-case 
    return newindx

coffee_exp.index = cleanUpCountryLabels(coffee_exp.index)
coffee_imp.index = cleanUpCountryLabels(coffee_imp.index)
coffee_prod.index = cleanUpCountryLabels(coffee_prod.index)
```

### 2. Standardizing column names to int for range and calculations

```python
#checks
coffee_exp.columns.is_unique
coffee_exp.columns.dtype
coffee_exp.columns
#cleaning
cols = coffee_exp.columns.astype(str).str.strip()
years = pd.to_numeric(cols[:-1])
coffee_exp.columns = list(years) + [cols[-1]] #make years a list so that it joins the columns list
coffee_exp.columns
```

```python
#checks
coffee_imp.columns.is_unique
coffee_imp.columns.dtype
coffee_imp.columns
#cleaning
cols = coffee_imp.columns.str.strip()
years = pd.to_numeric(cols[:-1])
coffee_imp.columns = list(years) + [cols[-1]]
coffee_imp.columns
```

```python
#checks
coffee_prod.columns.is_unique
coffee_prod.columns.inferred_type
coffee_prod.columns
#cleaning
cols = coffee_prod.columns.astype(str).str.strip() #only select year columns for range utility
years = pd.to_numeric(cols[1:-1].str[:4]) #use .str to remove mismatched string endings: '1990/91' -> 1990
coffee_prod.columns = [cols[0]] + list(years) + [cols[-1]]
coffee_prod.columns
```

## Part 4 – Business & Trade Questions

### 1. Largest change in production.

```python
#use prod_years from cleaning Part 3
prod = coffee_prod[prod_years]
first = prod.bfill(axis=1).iloc[:,0] #first valid date
last = prod.ffill(axis=1).iloc[:,-1]
change = last - first #used each country's first and last recorded values

best = change.abs().idxmax()
print(best, change[best])
```

```python
print(change.abs().nlargest(5))
print(prod.loc[best])
```

### 2. Trade balance.

```python
#no overlap in exports and imports --> no country does both
exp = coffee_exp["Total_export"]
imp = coffee_imp["Total_import"]
print(exp.index.intersection(imp.index))
```

```python
# no country appears in both exports and imports data sets, so each country's balance equals total exports or negative of its total imports
bal = exp.sub(imp, fill_value=0)
top_exp = bal.nlargest(5)
top_imp = bal.nsmallest(5)
print(top_exp)
print(top_imp)
```

#### 3. Production by coffee type.

```python
print(coffee_prod["Coffee type"].value_counts())

by_type = coffee_prod.groupby("Coffee type")["Total_production"].sum()
type_share = by_type / by_type.sum() #percentages
ratio = by_type / by_type.min() #x to 1 comparison, need for donut chart

print(by_type)
print(type_share.round(3))
print(ratio.round(2))
```

#### 4. Exporters with no production.

```python
print(coffee_prod.columns.tolist())
print(prod_years.tolist()[:3], len(prod_years))
```

```python
exporters = coffee_exp.index
producers = coffee_prod.index 
not_listed = exporters.difference(producers)
no_output = coffee_prod.index[coffee_prod[prod_years].isna().all(axis=1)]
silent_producers = exporters.intersection(no_output)

print(not_listed.tolist())
print(silent_producers.tolist())
```

## Part 5 – Visualizations

```python
global_exports = coffee_exp[exp_years].sum()    # adds down each column: one total per year

plt.figure(figsize=(10, 5))
plt.plot(global_exports.index, global_exports.values / 1e9, label="Total exports")
plt.title("Global Coffee Exports, 1990-2019")
plt.xlabel("Year")
plt.ylabel("Exports (billions, units as given in the data)")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 6))
plt.pie(type_share.values, autopct="%1.1f%%", wedgeprops=dict(width=0.4))
plt.title("Share of Global Coffee Production by Type")
plt.legend(type_share.index, title="Coffee type", loc="center left", bbox_to_anchor=(1, 0.5))
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 6))
ax.barh(top_exp.index, top_exp.values / 1e9, label="Net exporters")
ax.barh(top_imp.index, top_imp.values / 1e9, label="Net importers")
ax.set_title("Top 5 Net Exporters vs. Top 5 Net Importers of Coffee")
ax.set_xlabel("Trade balance (billions, exports minus imports)")
ax.set_ylabel("Country")
ax.legend()
plt.tight_layout()
plt.show()
```

Hello supply chain leadership at Roast Deez Beans Corp: Our main findings from the data analysis process indicate that coffee production is concentrated in a few countries, while its consuption is spread across many importers. The leading exporters are Brazil with 33.8 billion, Viet Nam with 24.9 billion, and Colombia with 19.9 billion. Leading in imports are the countries United States (42.5 billion), Germany (31.5 billion), and Italy (13.3 billion). There is a clean split in this data, since none of countries listed as exporters are also coffee importers. We have been able to analyze data from 1990 through 2019.

Brazil is not only the leading exporter, but has had the greatest increase in production of about 1.86 billion from 1990/91 to 2019/20. It is both the largest source of supply and is expanding the fastest. The world's dependence on one country is growing, not shrinking.

It should be noted that Brazil alone is 1/3 of all coffee exports, and the three leading exports are half of all coffee exports. A bad harvest in any of these countries could dramatically influence coffee availability and price increase due to scarcity. However, the top three exporters are also the top three producers, so partnering with these countries is well-supported by the data.

As for coffee types, only 3.4% exports were solely Robusta type. It may be worth checking whether this category can be marketed as a specialty commodity. The rest of the breakdown is 25.7% solely Arabica, and then 42.8% Arabica/Robusta blend and 28.1% Robusta/Arabica blend.

Note that our data analysis process treated zeroes as missing years, years are crop years (first year in the label), and no units were given with the data sets.
